## Recap:

We can plot graphs using lists or arrays. To do this we use the matplotlib (matrix plotting library) pyplot (python plotting) package. 

We use `import matplotlib.pyplot as plt` so we can use `plt` instead, and don't have to write it all out each time we use it.

we can use `plt.plot(x_data, y_data)` to plot a graph. This plots a plain blue line joining the dots. we can change the colour and style of the plot using a third argument, e.g. `plt.plot(x_data, y_data, 'xk')` plots 'x's for each point, and makes them black, as the letter 'k' represents black. 


```python
# 2
x_data = np.array([0,1,2,3,4,5])
y_data = np.array([-0.2, 0.4, 1.2, 1.6, 1.8, 2.7])

plt.plot(x_data, y_data) # gives a blue line joining the dots
plt.plot(x_data, y_data, 'xk') # gives a black x for each point
plt.plot(x_data, x_data/3, '-m'); # a bad line of best fit
```

Don't forget to give all graphs titles and axes labels!

```python
# 3
plt.plot(x_data, y_data, 'xk') # gives a black x for each point
plt.plot(x_data, x_data/3, '-m'); # a bad line of best fit
plt.title('Graph to show $x$ vs $y$')
plt.xlabel('$x$') 
plt.ylabel('$y$');
# the dollar signs $ allow latex notation to be used, so ^ gives superscript;
```

There are a range of colour codes you can use

- b : blue
- g : green
- r : red
- c : cyan
- m : magenta
- y : yellow
- k : black
- w : white

And markers

- "-" : straight line joining dots
- "." : dot
- "o" : circle
- "+" : plus
- "v" : triangle down
- "^" : triangle up
- "s" : square
- "p" : pentagon
- "8" : octagon
