# Plotting

We can plot graphs of these arrays using the `matplotlib.pyplot` library. 

First you will need to import this library

```python
import matplotlib.pyplot as plt
```

```python
# 25
# the semi colon stops the expression returning an object. 
# Try without and see what you get
plt.plot(arr, arr**2);  
# this plots points with x value my_arr, and y value my_arr**2
```

Graphs should have titles and axes labels

```python
# 26
plt.plot(arr, arr**2) 
plt.title('Graph to show array squared')
plt.xlabel('$arr$') 
plt.ylabel('$arr^2$'); # the dollar signs $ allows latex notation to be used, so ^ gives superscript;
```

You can plot more than one set of data on the same graph. This is useful for directly comparing data. 

```python
# 27
plt.plot(arr, arr**2) 
plt.plot(arr, arr*5)  
plt.title('Graph of $y$ vs $x$') 
# the dollar signs allow latex notation to be used
plt.xlabel('$x$') 
plt.ylabel('$y$') 
plt.grid(); #makes the graph more readable
```

We need to know which line is which!

 We add a label argument with each plot statement, so it knows which is which, then we use `plt.legend()` to make the key.

```python
plt.plot(arr, arr**2, label=r"$y = x^2$") 
#the label argument means we can use plt.legend() to see which line is which
plt.plot(arr, arr*5, label=r"$y = 5x$")
plt.title(r'Graph of $y$ vs $x$')
plt.xlabel('$x$') 
plt.ylabel('$y$') 
plt.legend() # this shows the key
plt.grid(); #makes the graph more readable
```
